```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/train.csv")

df.head()
```

```python
df.info()

df.describe()
```

```python
plt.hist(df["Age"], bins=100)
plt.title("distribuição das idades")
plt.xlabel("Idade")
plt.ylabel("Qunantidade")
plt.show()
```

```python
print("idade media das pessoas: ", df['Age'].mean())

print("mediana das idades das pessoas: ", df['Age'].median())

print("desvio padrão das idades: ", df['Age'].std())
```

```python
homens = df[df['Sex'] == 'male']

homens.describe()
```

```python
mulheres = df[df['Sex'] == 'female']

mulheres.describe()
```

```python
prob_homens_sobrevivem = (homens['Survived'] == 1).mean()

prob_mulheres_sobrevivem = (mulheres['Survived'] == 1).mean()

print(f"mulheres:, {prob_mulheres_sobrevivem:.2%}")
print(f"Homens:, {prob_homens_sobrevivem:.2%}")
```

```python
sobrevivencia_por_classe = df.groupby("Pclass")["Survived"].mean()
print(sobrevivencia_por_classe)
```

```python

df[df["Survived"] == 1]["Age"].hist(alpha=0.5, label="Sobreviveu")
df[df["Survived"] == 0]["Age"].hist(alpha=0.5, label="Não sobreviveu")
plt.legend()
plt.xlabel("Idade")
plt.ylabel("Quantidade")
plt.show()
```
