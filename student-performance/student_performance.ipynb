```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/StudentsPerformance.xls")

df.head()
```

```python
df.info()

df.describe()

df.isnull().sum()
```

```python
print("media das notas: ", df['math score'].mean()) # media das notas 

print("mediana das notas: ", df['math score'].median()) # mediana

print("devio padrão: ", df['math score'].std()) # desvio padrao
```

```python
plt.hist(df["writing score"], bins=100)
plt.title("distribuição do writing score")
plt.xlabel("Nota")
plt.ylabel("Frequência")
plt.show()
```

```python
sns.boxplot(data=df[["math score", "reading score"]])

plt.title("Matemática vs Leitura")
plt.show()
```

```python
probabilidade = (df['math score'] > 70).mean()

print(f"probabilidade de tirar mais que 70 em matematica {probabilidade:.2%}")
```
