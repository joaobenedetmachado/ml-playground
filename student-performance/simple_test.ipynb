```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/StudentsPerformance.xls")

df.head()
```

```python
def mode_mean_median(df):
    print(df.mean())
    print(df.median())
    print(df.mode())


def amp(df):
    print(df.max() - df.min())

```

```python
mode_mean_median(df['math score'])
mode_mean_median(df['reading score'])
mode_mean_median(df['writing score'])
```
