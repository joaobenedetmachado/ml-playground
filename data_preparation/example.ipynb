```python
import pandas as pd

df = pd.read_csv('data/data.csv')
```

Exploração de Dados
Frase resumida: "Explorar os dados mostra o que precisamos limpar ou transformar antes de treinar o modelo."

```python
df.info()

df.describe()
```

Limpeza de valores nulos e duplicatos
Frase resumida: "Dados limpos evitam que o modelo aprenda errado."

```python
df = df.drop_duplicates().fillna(df.mean(numeric_only=True))

print(df.describe())    
```

Normalização / Padronização
Frase resumida: "Normalizar dados deixa todos os números comparáveis."

```python
numeric_cols = ['Idade', 'Salário', 'Experiência']
df[numeric_cols] = (df[numeric_cols] - df[numeric_cols].mean()) / df[numeric_cols].std()

print(df.head())
```

O que significa os numeros de agora

basicamente eles deixaram de dados brutos e se tornaram dados relacionados entre si

quanto mais perto de 0, mais proximo da media da coluno

se é menor que 0, esta abaixo da media

se é maior que 0, esta acima

traducao de variveis "strings" para inteiros, que é como a ia de fato vai entender e tudo mais

```python
df['Departamento'] = df['Departamento'].map({'Vendas':0, 'Marketing':1})

print(df.head())
```

Divisão em Treino / Validação / Teste
Frase resumida: "Dividir os dados evita que o modelo só memorize os exemplos."

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df.drop('Salário', axis=1),  # Features (todas as colunas exceto 'Salário') pq é oq queremos prever
    df['Salário'],               # Target (o que queremos prever)
    test_size=0.2,               # 20% dos dados vão para teste
    random_state=42              # Garante que a divisão seja reproduzível
)

print("X_train:\n", X_train, "\n")
print("X_test:\n", X_test, "\n")
print("y_train:\n", y_train, "\n")
print("y_test:\n", y_test)
```

X_train: features de 4 linhas (80% dos dados), sem a coluna Salário.

X_test: features de 1 linha (20% dos dados), sem a coluna Salário.

y_train: salários correspondentes às linhas de X_train.

y_test: salário da linha que foi para teste.

Isso mostra que o train_test_split funcionou, separando o DataFrame original em treino e teste de forma reproduzível.
